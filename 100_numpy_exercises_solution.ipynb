4. How to find the memory size of any array

```python
arr = np.arange(5, 11)
# Return how many numbers in the array
items = np.size(arr)
# Return the size of one number (because all have same size)
size_of_item = arr.itemsize
memory_size = items * size_of_item
memory_size
```

11. Create a 3x3 identity matrix

```python
print(np.identity(3))
# or
print(np.eye(3, 3))
```

- **The difference between np.identity() and np.eye()**<br/>
In NumPy, `identity` creates a square identity matrix only, while `eye` creates an identity-like matrix of any shape.<br/>
    - np.identity(n) → always **n × n**<br/>
    - np.eye(n, m) → **n × m** (more flexible)

15. Create a 2d array with 1 on the border and 0 inside

```python
rows = int(input("Enter how many rows do you want: "))
columns = int(input("Enter how many columns do you want: "))
arr = np.ones((rows, columns), dtype=int)
arr[1:-1, 1:-1] = 0
print(f"So your 2D array with 1 on the border and 0 inside is:\n {arr}")
```
