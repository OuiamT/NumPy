### 5. Searching Arrays: 

```python
# 15
arr = np.array([-2,6,9,-9,6,-4,-3,3,4,-5])
arr[arr < 0]
```

```python
# 16
np.random.seed(42)
y = np.random.randint(1, 10, (3, 4))
y[y >= 5]
```

```python
# 17
arr1 = np.array([5,3,2,0,7,5,4])
arr2 = np.array([9,3,2,1,4,5,6])
np.where(arr1 == arr2)
```

```python
# 18
x = np.arange(1, 5) + 5
x
```

```python
# 19
np.random.seed(42)
arr = np.random.randint(1, 15, (3, 4))
arr[(arr > 5) & (arr < 10)]
```

```python
# 20
y = np.array([6,8,5,3,1,0,6,5])
print(f"The numbers higher than the elements mean are: {y[y > y.mean()]}")
```

### 8. Operation Math: 

```python
# 25
z = np.random.randint(1, 10, (4, 2))
# sum of all elements
print(np.sum(z))
# sum of all rows
print(np.sum(z, axis=1))
# sum of all columns
print(np.sum(z, axis=0))
```

```python
# 26
print(np.mean(z))
```
